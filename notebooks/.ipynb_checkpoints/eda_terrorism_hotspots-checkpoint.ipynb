## 1. Load & Inspect

```python
df = pd.read_csv("../data/raw/globalterrorismdb_0718dist.tar.bz2", compression="bz2")
df.info()
df.describe()
```

## Reframe & Inspect

```python
subset_cols = ["iyear", "country_txt", "region_txt", "city", "latitude",
               "longitude","success", "suicide", "attacktype1", "attacktype1_txt",
               "targtype1_txt", "targsubtype1_txt", "target1", "natlty1_txt", "gname",
               "gsubname", "nperps", "weaptype1_txt", "weapsubtype1_txt", "nkill", "nkillus"]
df_subset = df[subset_cols]
df_subset.info()
df_subset.describe()
```

### From description we can assume:
1. Columns "longitude", "latitude", "city", "nperps", "nkill" should be inspected for nan and unpredictable values
2. Column "nkill" need to be inspected for outliers

```python
print(df_subset.isna().sum())
print(df_subset.duplicated())
```

Cheking the percentage of nans

```python
print(df_subset["latitude"].isna().mean() * 100)
print(df_subset["longitude"].isna().mean() * 100)
```

Looking for unpredictable values

```python
df_subset["iyear"].unique()
```

```python
df_subset["region_txt"].unique()
```

```python
nan_kills = df_subset[df_subset["nkill"].isna()]
nan_kills_rate = (len(nan_kills) / len(df_subset)) * 100
print(nan_kills_rate)
print(nan_kills[["iyear", "attacktype1_txt", "success", "targtype1_txt", "nperps", "country_txt"]].head(10))
nan_kills_succ_rate = nan_kills["success"].sum() / len(nan_kills) * 100
print(nan_kills_succ_rate)
```

There are nearly 5.7% "nkill" without information, seems like it`s not an some error with data collection. By nan_kills_succ_rate we can see that mostly a terrorist attack took place, but no precise figures on the number of fatalities have been recorded. So was decided to make a subset of the data and analyze it seperatly.

```python
print(df_subset["nkill"].quantile([0.5, 0.9, 0.95, 0.99]))
print(df_subset["nkill"].value_counts().head(10))
q_99 = df_subset["nkill"].quantile( 0.99)
outliers = df_subset[df_subset["nkill"] > q_99]
```

There are 1% of events with much higher number of killed than in most of the data. This rows will be seperatly analyzed  as _High-Impact Incidents_.

```python
df_subset["attacktype1_txt"].unique()
unkn_atttype_rate = (df_subset["attacktype1_txt"] == "Unknown").sum() / len(df_subset) * 100
print(unkn_atttype_rate)
```
